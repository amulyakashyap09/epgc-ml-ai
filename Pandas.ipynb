```python
# Merge between two data frame

players = ['p1', 'p2', 'p3']
points = [8, 10, 12]
games = ['g1', 'g2', 'g3']

df1 = pd.DataFrame({
    'Players': players,
    'Points': points,
    'Games': games,
})
df1
```

```python
# Merge between two data frame

players = ['p1', 'p5', 'p6']
powers = ['sprint', 'jump', 'lick']
games = ['g1', 'g2', 'g3']

df2 = pd.DataFrame({
    'Players': players,
    'Powers': powers,
    'Games': games,
})
df2
```

```python
# now merge df1 and df2 using INNER MERGE (DEFAULT) based on any common column

df1.merge(df2, on='Players', how='inner')
```

```python
# now merge df1 and df2 using LEFT MERGE based on any common column

df1.merge(df2, on='Players', how='left')
```

```python
# now merge df1 and df2 using RIGHT MERGE based on any common column

df1.merge(df2, on='Players', how='right')
```

# Join using panda

## merge - atleast 1 common column between two datasets
## join - atleast 1 common index between two datasets

```python
# Merge between two data frame

players = ['p1', 'p5', 'p6']
powers = ['sprint', 'jump', 'lick']
games = ['g1', 'g2', 'g3']

df3 = pd.DataFrame({
    'Playerss': players,
    'Powerss': powers,
    'Gamess': games,
}, index=['L1', 'L2', 'L3'])
df3

players = ['p1', 'p3', 'p4']
powers = ['sprint', 'jump', 'lick']
games = ['g1', 'g2', 'g3']

df4 = pd.DataFrame({
    'Players': players,
    'Powers': powers,
    'Games': games,
}, index=['L1', 'L2', 'L3'])
df4
```

```python
df3.join(df4, how='inner')
```

```python
# Left Join

df3.join(df4, how='left')
```

```python
# Left Join

df3.join(df4, how='right')
```
