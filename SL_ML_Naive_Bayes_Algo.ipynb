## Understanding Conditional Probability

- Calculate the probability of second Event (Event B) given that the first event (event A) has already happend

**For example**
- Drawing a second ACE card from a deck given we got the first ace
- Finding the probability of having a disease given you were tested positive
- Finding the probability of liking Game of Thrones given that person like the fiction genre

## Definition and Formula

- Define two events ...
    - Event A is the probability of the event we're trying to calculate
    - Event B is the condition that we know or the event that has happend
- **Conditional Probability = P(A|B)**
    - The probability of the occurrence of event A given that B has already happned
    - P(A|B) = P(A ∩ B) / P(B) = Probability of the occurence of both events A and B / Probability of A
    - **P(A ∩ B) = P(B) * P(A|B)** ✅ Correct
        - ⚠️**P(A ∩ B) = P(A) * P(B) is only true when A and B are independent events.**
    - **P(A|B) = (P(B) * P(A|B)) / P(B) = Probability of the occurence of both events A and B / Probability of A**

**Example 1**: 
Suppose you have two jars having 6 marbles, 3 red and 3 black. What is the probability of getting a black marble second time given that first was black marble drawn.

- P(A) = Probability of Getting black marble **first** time = no of black marble / total marbles = 3/6
- P(B|A) = Probability of Getting black marble **second** time = left black marbles after event A / left all marble after event A = 2/5
- P(A|B) = (P(B) * P(B|A)) / P(B) = ((3/6) * (2/5)) / P(2/5) = 0.4 = 40%
    - **P(A|B) = What I want given what I know = What I know given what I want × probability of what I want ÷ probability of what I know**
        - A|B = A given B
    - **P(B|A) = What I know given what I want**
        - B|A = B given A
    - **P(A) = probability of what I want**
    - **P(B) = probability of what I know**
- **Bayes Theorem** = P(A|B) = (P(B|A) * P(A))/P(B)


## Bayes Theorem : Use Case

**Question :** Find out a patientsprobability of having a liver disease if they are an alcoholic
Answer:
1. A event "Patient has a liver disease" Past data tells you that 10% of patients entering your clinic has liver disease.
    2. **P(A) = 0.10**
3. B test that "Patient is an alcoholic". Five percent of the clinic's patient are alcoholic.
    4. **P(B) = 0.05**
5. **B|A** = Also, Out of all Person having liver disease, 7% are alcoholics = 0.07
6. Bayes Theorem P(A|B) = (0.07 * 0.1)/0.05 = 0.14
