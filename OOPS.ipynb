# Inhertiance [inheriting the properties from parent class/pbject

1. Single Inheritance - can only inherit from one parent class
2. Multiple Inheritance - can inherit from multiple parent classes
3. Multi level - one class inherits from another class which further inherits from another class
4. Heirarichal Inheritance - one parent class is inherited by mulitple child classes
5. Hybrid - combination of any two kinds of inheritance

```python
# Single Inheritance

#Syntax: class(parent_class_name):

class fruit:
    def __init__(self):
        print('I am a fruit')

class mango(fruit):
    def __init__(self):
        print('I am a mango fruit!')

myFruit = mango()
    
```

# Multiple Inheritance

## Syntax: class(parent_class_A, parent_class_B):

### Note: Inheriting from multiple classes can cause issue

```python
# Multilevel Inheritance

## Syntax:

class A:
    x = 1
class B(A):
    pass
class C(B):
    pass

cobj = C()
print(cobj.x)
```

# Heirarichal Inheritance

## Syntax: class A:
## Syntax: class B(A):
## Syntax: class C(A):
## Syntax: class D(A):

## one parent class is inherited by mulitple child classes

```python
# Hybrid Inheritance

## Syntax:

class A:
    x = 1
class B(A): # hierarichal inheritance
    pass
class C(A): # hierarichal inheritance
    pass
class D(B,C): # multiple inheritance
    pass

cobj = D()
print(cobj.x)
```

# Inheritance SUPER function

It is a special method to call/invoke parent class

```python
# SUPER FUNCTION
class Vehicle:
    def start(self):
        print('starting engine')
    def stop(self):
        print('stopping engine')

class Twowheeler(Vehicle):
    def say(self):
        super().start()
        print('I have two wheeler vehicle')
        super().stop()

x = Twowheeler()
x.say()
```

```python
# POLYMORPHISM - BASIC

class Dog:
    def sound(self):
        print("Dog barks")

class Cat:
    def sound(self):
        print("Cat meows")

animals = [Dog(), Cat()]
for i in animals:
    i.sound()
```
