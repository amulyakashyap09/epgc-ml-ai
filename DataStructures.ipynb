# Tuples - immutable data structure, which safe for get operation, can be concatenated to a new tuple but cannot be modified
# Though tuple cannot be modified but object inside tuple can be modified

# What you cannot do with a tuple
❌ Change an element
❌ Add new elements
❌ Remove elements

# What you can do with a tuple

✅ Get an element from tuple
✅ repeat elements in tuple
✅ Slicing elements from tuple
✅ Concat tuples

```python
tupl = ('a', 'b', 'c')
print(tupl)
```

```python
#indexing of tuple #fetching an element from tuple 
tupl = ('a', 'b', 'c')
print(tupl[0])
```

```python
tupl = ('a', 'b', 'c')
# tupl[0] = 'x' 
# cannot be modified , this will give error
# print(tupl)
```

```python
tupl = ('a', ['b', 'c'])
tupl[1][0] = 'x'

# thought tuple cannot be modified but object inside tuple can be modified
print(tupl[1])
```

```python
tupl = ('a', 'b', 'c')
tupl*2
```

```python
tupl = (1, 2, 3, 4, 5)
tupl[2:4]
```

```python
#concating tuples

tupl = (1, 2, 3, 4, 5)
tupl1 = (1, 2, 3, 4, 5)
tupl += tupl1

print(tupl)
```

# Dictionaries

```python
myDict = {'a': 1, 'b': 2, 'c':3}
print(myDict)

myMixDict = {'a': 1, 'b': ['a', 'b'], 'c':3.14}
print(myMixDict)

myPairDict = dict([('a', 1), ('b', 2)])
print(myPairDict)

print(myDict['a'])

print(len(myDict))

print(myDict.values())

print(myDict.keys())

print(myDict.items())
```

```python
mySet = {1, 2, 3} #unordered unique items collection
print(mySet)

mySet1 = {1, 2, 3}
mySet2 = {3, 4, 5}

print(mySet1 | mySet2) #union
print(mySet1 & mySet2) #intersection
print(mySet2 - mySet1) #difference
```

```python
i=19

match i:
    case 1:
        print(i)
    case 2:
        print(i)
    case 3:
        print(i)
    case _:
        print("default: ", i)
```

```python
num = 100

def checkPrime(n):
    is_prime = True
    if n==1:
        return True
    for i in range(2, n):
        if n%i == 0:
            is_prime = False
            break
    return is_prime

for i in range(1, num+1):
    output = checkPrime(i)
    print(i, "Is Prime ", output)
```
