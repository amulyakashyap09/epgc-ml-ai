# Probability

## Probability-based sampling 

It is a sampling method in statistics where **every member of the population has a known, non-zero chance of being selected**.

## Simple Definition

If selection is based on **randomness**, it is probability sampling.

Example:
Suppose a company has 10,000 users and wants feedback from 500 users.

If every user has a measurable chance of being chosen using a random process, that is probability-based sampling.

## Main Types of Probability Sampling

### 1. Simple Random Sampling

Every individual has an equal chance of selection.

Example:
Picking 100 employee IDs randomly from a database.

---

### 2. Systematic Sampling

Choose every k-th item after a random start.

Example:
Survey every 20th customer entering a store.

---

### 3. Stratified Sampling

Divide population into groups (strata) and randomly sample from each group.

Example:
Selecting students from each class year separately:

* 1st year
* 2nd year
* 3rd year

This ensures balanced representation.

---

### 4. Cluster Sampling

Divide population into clusters and randomly choose whole clusters.

Example:
Randomly selecting 5 schools and surveying all students in those schools.

---

## Why It Matters

Probability sampling:

* Reduces bias
* Produces more reliable conclusions
* Allows statistical inference about the population

Non-probability sampling (like convenience sampling) does not guarantee this.

# 1. Marginal Probability

Marginal probability means the probability of a **single event occurring**, ignoring other variables.

Example:

* Probability that a student is male
* Probability that a customer buys a product

Suppose:

|        | Buys Product | Doesn't Buy | Total |
| ------ | ------------ | ----------- | ----- |
| Male   | 30           | 20          | 50    |
| Female | 10           | 40          | 50    |
| Total  | 40           | 60          | 100   |

Probability that a randomly selected person is male:

[
P(\text{Male}) = \frac{50}{100} = 0.5
]

This is called **marginal** because we look at the row/column totals (margins).

P(A)=\sum_{B} P(A,B)

Meaning:
To get marginal probability of A, sum over all possible values of B.

---

# 2. Joint Probability

Joint probability means the probability of **two events happening together**.

Example:

* Person is Male AND buys product

From the table:

[
P(\text{Male AND Buy}) = \frac{30}{100} = 0.3
]

This represents the intersection of two events.

# 3. Conditional Probability

Conditional probability means:
Probability of event A happening **given that** event B has already happened.

Example:
Probability that a person buys product GIVEN that the person is male.

Among males:

* 30 bought
* Total males = 50

So:

[
P(\text{Buy} \mid \text{Male}) = \frac{30}{50} = 0.6
]

Conditional probability comes from joint probability:

# Central Limit Theorem (CLT)

## Definition: If you repeatedly take samples and compute their averages, those averages tend to follow a Normal (Gaussian) distribution — even if the original data is NOT normal.

**Core Idea**

***Suppose original data is:*** weird, skewed, random, non-Gaussian

***If***: sample size is reasonably large

***then:*** distribution of sample means→Normal distribution

### Important Distinction

CLT talks about: **Distribution of SAMPLE MEANS NOT original data.**

## Why CLT Is So Important

### Because it allows us to:

1. estimate population behavior
2. do hypothesis testing
3. build confidence intervals
4. use Gaussian assumptions

### Even when raw data is not Normal.

#### Example: Elections

You cannot ask every voter. So:

1. sample 1000 people
2. compute average preference

#### CLT helps justify:

**polling accuracy**

#### Example: Machine Learning

1. Suppose: millions of user interactions
2. Average CTR, average loss, average reward: often become approximately Normal.

## Key Conditions for CLT

1. Samples should be independent
2. Sample size should be sufficiently large

## What Happens As Sample Size Increases

1. Small Sample - Means distribution still messy.
2. Larger Sample - More bell-shaped.
3. Very Large Sample - Almost perfectly Gaussian.

**Instead of estimating a single number, We estimate a range.**

### Example

Take a random normal sample of size 100, and calculate the lower and upper confidence interval with interval value 95%

````python
import numpy as np
import seaborn as sns
import scipy.stats as st
import statistics as s
import matplotlib.pyplot as plt
%matplotlib inline

x = np.random.normal(size=100)
sample_mean = np.mean(x)
sample_stdev = s.stdev(x)
std_err = st.sem(x)
# Z = st.norm.ppf(1-0.05)
Z = st.norm.ppf(1 - 0.05/2)

lowerCi = sample_mean - (Z * std_err)
upperCi = sample_mean + (Z * std_err)

plt.figure(figsize=(15,5))
# sns.displot(x)
sns.kdeplot(x)
plt.axvline(x=lowerCi, color='red')
plt.axvline(x=upperCi, color='red')
plt.axvline(x=sample_mean, color='green')
````

# T-Test

**A t-test is a statistical test used to determine whether the difference between means is statistically significant or could have occurred by random chance.**

**In simple words:** "I observed a difference in averages. Is this difference real, or just sampling noise?"

## Why do we need a t-test?

Suppose you measure the heights of 20 students and get:

1. Sample mean = 172 cm
2. Sample standard deviation = 8 cm

You want to know:

"Is the true population mean height different from 170 cm?"

**Since you only have a sample, not the entire population, there's uncertainty. The t-test quantifies that uncertainty.**

## Assumptions in T-test
- The samples are independed
- Homegeneity in sample variances
- The samples are normally distributed

## Types of t-tests

### One-Sample t-test

Tests whether a sample mean differs from a known value.

Example:

- Average IQ = 100?
- Average salary = ₹10 lakh?

### Two-Sample t-test

Compares means of two independent groups.

Example:

- Male vs Female salary
- Control group vs Treatment group

### Paired t-test

Compares measurements before and after an intervention on the same subjects.

Example:

- Weight before diet vs after diet
- Blood pressure before medicine vs after medicine

## Understanding the p-value

After performing a t-test, you get:

```
TtestResult(
    statistic=2.45,
    pvalue=0.018
)
```

1. The p-value answers: If the null hypothesis were true, how likely is it to observe a result this extreme?
2. Common rule:
- p < 0.05 → statistically significant
- p ≥ 0.05 → insufficient evidence

**Example: p = 0.018, means there is only a 1.8% chance of seeing such a difference if there were actually no effect.**

## Why "t" instead of "z"?

You learned confidence intervals using a Z-score (1.96).

Use:

- Z-test when population standard deviation (σ) is known.
- t-test when population standard deviation is unknown.

**In real-world data science, σ is almost never known, so t-tests are much more common.**

As sample size increases:

t-distribution → normal distribution

For very large samples, t and z give almost identical results.

## One-Tailed Test

You're asking:

"Is the mean specifically higher?"
or
"Is the mean specifically lower?"

**Only one direction matters.**


## Two-Tailed Test

You're asking:

"Is the mean different?"

It could be either higher or lower.

Hypotheses

For a population mean:

- H0  ​:μ=100
- Ha  ​:μ≠100

Notice the ≠.

**You're interested in deviations in both directions.**

# F-Test

It is statistical test which compares the variances of the two population

## Assumpitions
- Data has to be normally distributed
- Data has to be independent

````python
# F-Test

# We will calculate the variances of two samples and compute the f-statistics and p-value
# to gather the statistical evidence to reject the null hypothesis

import scipy
from random import sample

samplex = sample(range(0, 100), 30)
sampley = sample(range(0, 100), 30)
f = np.var(samplex) / np.var(sampley)
p = 1 - scipy.stats.f.cdf(f, len(samplex)-1, len(samplex)-1)
print('f : ', f)
print('p : ', p)

# Since p > 0.05 , hence we fail to reject the null hypothesis.
````

# ANOVA - Analysis of Variance

ANOVA (Analysis of Variance) is a statistical test used to determine whether the means of three or more groups are significantly different from each other.

**Although its name mentions "variance," its primary purpose is to compare means.**

## Why do we need ANOVA?

Suppose you want to compare the average marks of students taught using three different methods.

| Teaching Method | Average Marks |
| --------------- | ------------- |
| Method A        | 75            |
| Method B        | 82            |
| Method C        | 78            |

**You could ask: Are these differences due to chance, or is at least one teaching method genuinely different?**

ANOVA answers this question.

### Why not use multiple t-tests?

As the number of group increases , your number of t-tests also increases. Example: you have four groups A, B, C, D then t-tests will AvB, AvC, AvD,BvC, BvD, CvD that becomes 6 tests to perform. **ANOVA performs one overall test instead.**

## Assumptions
- Independent Samples
- All population have common variance
- Samples are drawn from normal distribution population

## Hypothesis
The hypotheses

For k groups:

**Null hypothesis (H0)**: All population means are equal.

**Alternative hypothesis (H1)**: At least one group's mean is different.

**It does not tell you which one—only that not all means are equal.**

## One Way ANOVA

We have to check the effect of 4 diffrent performance enhancer in the electric vehicle is same or not
- Null Hypothesis: The performance averages are equal.
- Alternate Hypothesis: The performance averages are not equal.

````python
## One Way ANOVA
# F-Test

from scipy.stats import f_oneway
from random import sample

sampleu = sample(range(0, 100), 30)
samplex = sample(range(0, 95), 30)
sampley = sample(range(0, 120), 30)
samplez = sample(range(0, 145), 30)

f_oneway(sampleu, samplex, sampley, samplez)
````

## Two Way ANOVA

We have to check how 2 factors will affect the response variable
- Null Hypothesis: There is no significance of the two factors on the response variable.
- Alternate Hypothesis: There is a significance of the two factors on the response variable.

````python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

x = {
    # 'Lectures': np.repeat(["Daily", "Weekly"], 20),
    # 'Tuition': np.repeat(["Daily", "Weekly"], 20),
    'Lectures': ['Daily']*5 + ['Daily']*5 + ['Weekly']*5 + ['Weekly']*5,
    'Tuition':   ['Daily']*5 + ['Weekly']*5 + ['Daily']*5 + ['Weekly']*5,
    'Marks': [78,80,79,81,77,82,84,83,85,81,70,72,71,73,69,75,76,74,77,73]
}

data = pd.DataFrame(x)
model = ols('Marks ~ C(Lectures) + C(Tuition) + C(Lectures):C(Tuition)', data=data).fit()
sm.stats.anova_lm(model, typ=2)
````
