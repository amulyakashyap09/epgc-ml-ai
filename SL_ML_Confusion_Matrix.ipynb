# Confusion Matrix

- The confusion Matrix shows the way your classification model is confused when it makes the prediction.
- It is summary of prediction results on a classification problem
- Key to Confusion Matrix: Count the value of corrct and incorrect predictions (broken down by class)

## How to calculate the Confusion Matrix


- You need a test dataset or a validation dataset with expected outcome values
- Make prediction for each row in your dataset
- Form the exptected outcomes and predictions, **count**:
    - The number of correct predictions for each class
    - The number of incorrect predictions for each class

# Confusion Matrix in Python

```python
from sklearn.metrics import confusion_matrix

expected_values = [1, 1, 0, 1, 0, 0, 1, 0, 0, 0]
predicted_values = [1, 0, 0, 1, 0, 0, 1, 0, 1, 1]

results = confusion_matrix(expected_values, predicted_values)
print(results)

"""
[[4 2]
 [1 3]]

 4 - 0 predicted as 0 - 4 times # correct prediction
 2 - 0 predicted as 1 - 2 times # incorrect prediction
 1 - 1 predicted as 0 - 1 time  # incorrect prediction
 3 - 1 predicted as 1 - 3 times # correct prediction

 Total prediction made = 10
 Total Correct predictions = 7 times
 Total Incorrect predictions = 3 times

 Accuracy = 70%
"""
```
