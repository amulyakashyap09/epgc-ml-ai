# Matplotlib

```python
import matplotlib.pyplot as plt
%matplotlib inline

import pandas as pd

cars = pd.read_csv('./datasets/cars.csv')
y1 = cars['hp']
x = range(32)

plt.plot(x, y1)
```

```python
import matplotlib.pyplot as plt
%matplotlib inline

import pandas as pd

cars = pd.read_csv('./datasets/cars.csv')
y1 = cars['hp']
y2 = cars['disp']
x = range(32)

plt.plot(x, y1)
plt.plot(x, y2)
plt.legend()
```

```python
import matplotlib.pyplot as plt
%matplotlib inline

import pandas as pd

cars = pd.read_csv('./datasets/cars.csv')
y1 = cars['hp']
y2 = cars['disp']
x = range(32)

plt.stackplot(x, y1, y2,
              colors=['purple', 'black'],
              alpha=0.6,
              labels=['hp', 'disp'])
```
