```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

#boston_dataset
boston = pd.read_csv("./datasets/boston.csv")
boston.head()
```

```python
#plot the chart
plt.scatter(x=boston['rm'], y=boston['medv'], color='brown')
plt.xlabel('Avg. no. of rooms per dwelling')
plt.ylabel('Median Value of home')
# as avg no of rooms increase median price of homes also increase
# indirectly size of room will increase too
```

```python
x = pd.DataFrame(boston['rm']) # single feature or single independent variable
y = pd.DataFrame(boston['medv']) # target
```

```python
# splitting into train and test

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20) # 20% records in test, 80% in training
```

```python
from sklearn.tree import DecisionTreeRegressor
regressor = DecisionTreeRegressor()
regressor.fit(x_train, y_train)
```

```python
# predicting the value
y_pred = regressor.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error
mse = mean_squared_error(y_pred, y_test)
rmse = np.sqrt(mse)
rmse
```

```python
# Now Introducing more Independent Variables in the models

x = pd.DataFrame(boston[['rm', 'lstat', 'age']]) # multiple features or multiple independent variables
y = pd.DataFrame(boston['medv']) # target or dependent variable

# splitting into train and test

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.30) # 30% records in test, 70% in training
```

```python
from sklearn.tree import DecisionTreeRegressor
regressor = DecisionTreeRegressor()
regressor.fit(x_train, y_train)
```

```python
y_pred = regressor.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error
mse = mean_squared_error(y_pred, y_test)
rmse = np.sqrt(mse)
rmse


# second model is better than first model and it produces less error
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

#boston_dataset
iris = pd.read_csv("./datasets/iris.csv")
iris.head()
```

```python
x = pd.DataFrame(iris[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']])
y = iris['species']

print(x.shape)
print(y.shape)
```

```python
# splitting into train and test

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.30) # 30% records in test, 70% in training
```

```python
from sklearn.tree import DecisionTreeClassifier
classifier = DecisionTreeClassifier()
classifier.fit(x_train, y_train)
```

```python
y_pred = classifier.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix
print(confusion_matrix(y_test, y_pred))

# left diagnoal values are correctly predicted values
# 3 is only wrongly classified value rest are correctly classified
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test, y_pred))
```
