# Machine Learning

## Types
- Supervise Learning
- Unsupervise Learning
- Reinforcement Learning

## Supervise Learning - Labelled training dataset and machine learns on that label.
- Example - Machine is taught about apple by showing apple

### Use Case 1 : Spam Classifier

How do you think your mail classifies as spam or not ? Works on base of filter (new tech, feedback by user)
- Text filter
- Eliminate thread based on sender history whether reported or not

#### Text Filter - alogrithm works on word phrases most often uses in the spam email, free ,won, lottery, bitcoin, etc
#### Client Filter - Client identity and history , if user has send lot of bulk email , several email reported as spam by others, text filter already raised a flag. Then it is market as spam or blacklisted

### Use Case 2 : Fingerprint Analysis

Saves your finger print in the machine and scans through all of data when scanned again

## Unsupervised Learning

It does not have any label, machine identifies based on the cluster or groups in the data. It finds the pattern in the images and groups them in clusters. It can divide the images based on pattern and then process in cluster.

#### Use Case 1: Voice Based Personal Assistant
#### Use Case 2: Netflix Platform Recommendation
- Implicit - you don't tell anything machine it chooses from your action
- Explicit - you tell machine by polling etc

## Reinforcement Learning

When machine takes right decision +1 point awarded and -1 when wrong. By iterating this multiple times, we train the machine. 

#### Use Case: Self Driving Cars

IoT, Sensors, Camera, Cloud Computing, Software Algorithm

All data collected by cars is use to train the self driving car.

# What can you do with Machine Learning Algorithm ?

## what type of questions can be asked to machine with:
- Classification Algorithm - **where we classify the object**. **category is predicted using data**
    - Is the person male or female ?
    - Is this dog or cat ?
    - Is the mail spam or not ?
- Anamoly Detection Algorithm - **Identify using unusual data points**
    - Is there any fruad in this transaction ?
    - Is someone trying to hack our network ?
- Clustering Algorithm - **Groups data based on some condition**
    - Which type of house lies in this segment ?
    - What type of customer buys this data ?
    - Recommendation Engine
- Regression Algorithm - **Groups data based on some condition**
    - What is the market value of this house ?
    - Is it going to rain tomorrow ?

```python
'''Importing all required libs'''
import pandas
from pandas.plotting import scatter_matrix
import matplotlib.pyplot as plt

from sklearn import model_selection
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC


url = './datasets/iris.csv'
names = ['sepal_length','sepal_width','petal_length','petal_width','species']
dataset = pandas.read_csv(url, names=names)

print(dataset.shape)
```
